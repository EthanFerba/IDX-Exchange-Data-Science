After these changes we can already see that the data for the model is already looking better, next we need to change the data types and then focus on handling the rest of the NA values within these feilds. Weather we are going to drop the rows that have na, or use Imputation

```python
# Keep only rows where parking is 10 spaces or fewer
rdf = rdf[rdf['ParkingTotal'] <= 10]
# Fill any remaining blank parking fields with the median
rdf['ParkingTotal'] = rdf['ParkingTotal'].fillna(rdf['ParkingTotal'].median())
rdf.isna().mean() * 100
```

```python
rdf = rdf.dropna(subset = ['LotSizeAcres'])
rdf = rdf.dropna(subset = ['LotSizeSquareFeet'])
rdf = rdf.dropna(subset = ['YearBuilt'])
rdf = rdf.dropna(subset = ['BathroomsTotalInteger'])
rdf = rdf.dropna(subset = ['City'])
rdf = rdf.dropna(subset = ['PurchaseContractDate'])



rdf.isna().mean() * 100
```

We can see for the garage spaces vs attachedgarage YN we can see that they should some what be related but the numbers do not connect. To fix this, I am going to cross refrence then and if they have a value in garage spaces, then in the attached garage YN field with return a yes

```python
rdf.loc[(rdf['GarageSpaces'] > 0) & (rdf['AttachedGarageYN'].isna()), 'AttachedGarageYN'] = 1
rdf.loc[(rdf['AttachedGarageYN'] == 0) & (rdf['GarageSpaces'].isna()), 'GarageSpaces'] = 0
# do a final sweep through to fill in nas with 0 in both columns if they did not get caught.
rdf['GarageSpaces'] = rdf['GarageSpaces'].fillna(0)
rdf['AttachedGarageYN'] = rdf['AttachedGarageYN'].fillna(0)
rdf.isna().mean() * 100
```

Next move towards new contstruction

```python
rdf = rdf.dropna(subset= ['NewConstructionYN'])
```

work on changing the YN answers to binary

```python
boolean_cols = ["AttachedGarageYN", "NewConstructionYN", "BasementYN"]

for col in boolean_cols:
    rdf[col] = rdf[col].fillna(False).replace("", False).astype(int)
```

```python

boolean_cols = ["AttachedGarageYN", "NewConstructionYN", "BasementYN"]

for col in boolean_cols:
    print(f"\n{col}:")
    print(rdf[col].value_counts(dropna=False))
```

```python
rdf['Stories'] = rdf['Stories'].fillna(rdf['Stories'].median())
levels_mode = rdf['Levels'].mode()[0]
rdf['Levels'] = rdf['Levels'].fillna(levels_mode)
# fill all the NAs in highschool with unknown, I think this field will play a huge part in determing the price
rdf['HighSchoolDistrict'] = rdf['HighSchoolDistrict'].fillna('Unknown')
#final check for no Missing values
rdf.isna().mean() * 100
```

Next we are going to make the final touches in changing the encoding of the data fields.

```python
rdf['PurchaseContractDate'] = pd.to_datetime(rdf['PurchaseContractDate'], errors='coerce')
rdf.dtypes
```

```python

# Create a list of flooring materials, excluding missing values and SeeRemarks
materials = sorted({
    material.strip()
    for value in rdf["Flooring"].dropna()
    for material in value.split(",")
    if material.strip() and material.strip() != "SeeRemarks"
})

# Create a binary column for each flooring material
for material in materials:
    rdf["Flooring_" + material] = rdf["Flooring"].apply(
        lambda x: int(material in [m.strip() for m in x.split(",")])
        if isinstance(x, str) else 0
    )

# Track missing flooring information
rdf["Flooring_Missing"] = rdf["Flooring"].isna().astype(int)

# Track listings where flooring details are referred to elsewhere
rdf["Flooring_SeeRemarks"] = rdf["Flooring"].fillna("").apply(
    lambda x: int("SeeRemarks" in [m.strip() for m in x.split(",")])
)

# Remove the original text column after encoding
rdf = rdf.drop(columns=["Flooring"])
```

Finally add the splitting of data for the train and testing scenarios

```python

def create_temporal_split(df, training_months_x):

    df = df.copy()
    df["CloseDate"] = pd.to_datetime(df["CloseDate"], errors="coerce")

    max_close_date = df["CloseDate"].max()
    test_start_date = max_close_date.to_period("M").start_time
    test_end_date = test_start_date + pd.DateOffset(months=1)

    train_start_date = test_start_date - pd.DateOffset(
        months=training_months_x
    )

    train_mask = (
        (df["CloseDate"] >= train_start_date)
        & (df["CloseDate"] < test_start_date)
    )

    test_mask = (
        (df["CloseDate"] >= test_start_date)
        & (df["CloseDate"] < test_end_date)
    )

    df_train = df[train_mask].copy()
    df_test = df[test_mask].copy()

    print(f"--- Split Summary for X = {training_months_x} Months ---")
    print(
        f"Train: {train_start_date.date()} to "
        f"{(test_start_date - pd.Timedelta(days=1)).date()} "
        f"({len(df_train)} rows)"
    )
    print(
        f"Test:  {test_start_date.date()} to "
        f"{(test_end_date - pd.Timedelta(days=1)).date()} "
        f"({len(df_test)} rows)"
    )
    print(f"Missing CloseDate rows: {df['CloseDate'].isna().sum()}\n")

    return df_train, df_test


window_options = [3, 6, 12, 24]

for X in window_options:
    train_set, test_set = create_temporal_split(
        rdf, training_months_x=X
    )
```
